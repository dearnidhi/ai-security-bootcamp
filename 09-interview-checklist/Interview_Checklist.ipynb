# 09 - Final Interview Checklist

Answer every question out loud, 2 minutes max, without notes. Whatever you hesitate on, go back to that module only.

## Module 02 - OWASP + Prompt Injection
- [ ] What is the OWASP Top 10 for LLM Applications? Name as many as you can.
- [ ] What is prompt injection?
- [ ] Direct vs indirect prompt injection?
- [ ] How do you mitigate prompt injection?
- [ ] What is jailbreaking?
- [ ] What is system prompt leakage?
- [ ] How do you prevent PII leakage?
- [ ] What is insecure output handling?

## Module 03 (Part 1) / 04 - Guardrails + LLM Gateways
- [ ] What are input/output guardrails, and how would you design them for a chatbot?
- [ ] What is a topical guardrail and why would you need one?
- [ ] How do you restrict which tools an agent is allowed to call?
- [ ] What is an LLM gateway and why would you use one?
- [ ] How does a gateway improve reliability when a provider is down or rate-limited?
- [ ] Why would you cache LLM responses, and what are the risks?

## Module 03 (Part 2) - Guardrail Frameworks
- [ ] Guardrails AI vs NeMo Guardrails - how do they differ?
- [ ] What are the on_fail actions (block, fix, log only)?
- [ ] Rule-based vs LLM-based rails - what are the trade-offs?
- [ ] Why redact only the secret instead of replacing the whole answer? (false positives)
- [ ] How do input and output guards work together as defense in depth?

## Module 05 - Agentic AI Security
- [ ] What security risks exist in AI agents?
- [ ] What is excessive agency?
- [ ] How do you secure tool/function calling?
- [ ] How do you prevent an agent from performing unauthorized actions?
- [ ] What is memory poisoning?
- [ ] Why is least privilege important for agents?

## Module 06 - LLM Evaluation + Red Teaming
- [ ] How would you red-team an LLM application?
- [ ] What is a faithfulness evaluator and why does it matter for RAG systems?
- [ ] How do you detect hallucination in an LLM's output?
- [ ] What are Garak, PyRIT, and Promptfoo used for?

## Module 07 - RAG Security
- [ ] What is RAG poisoning?
- [ ] How do you secure a RAG pipeline end to end?
- [ ] How do you enforce document-level access control?
- [ ] How would you detect a poisoned document in a retrieval corpus?

## Module 08 - Capstone + Secure Architecture
- [ ] How would you design a secure GenAI application end to end?
- [ ] How would you monitor security issues in production?
- [ ] Walk me through this project: what attack does it demonstrate and how did you defend against it?

## Module 10 - Secure AI Architecture (BOLA / tenant isolation)
- [ ] What is BOLA/IDOR, and how does it apply to an AI backend, not just a REST API?
- [ ] Why is checking "is this API key valid" not enough?
- [ ] How do you enforce tenant isolation in a multi-tenant RAG or agent system?
- [ ] Why should an audit log be access-controlled and scoped?

## Module 11 - Red Teaming with PyRIT
- [ ] What are the building blocks of PyRIT (target, attack, converter, scorer)?
- [ ] What is Attack Success Rate, and how would you use it in CI?
- [ ] What is XPIA (indirect prompt injection)? Give an example.
- [ ] Why do encoding attacks (Base64, leetspeak) work against input filters?
- [ ] PyRIT vs Garak vs Promptfoo - when would you use each?
