# AI Security Course

Senior AI/ML Engineer interview prep, built around your own repo: `github.com/dearnidhi/ai-security-series`

All API calls in this course use **Groq only** (via the shared root `.env`). No other provider is needed for any module.

## One-time setup (shared by every module)

```bash
# from this folder (21-AI security)
python -m venv venv
venv\Scripts\activate
pip install -r requirements.txt
```

The `.env` file at this folder's root already has `GROQ_API_KEY` set. `load_dotenv()` in every module's code searches upward automatically, so you do **not** need to copy `.env` into each module folder - one root `.env` covers all of them.

Module **11** (PyRIT) and the frameworks part of **03** (Guardrails AI, NeMo Guardrails) install big libraries, so `pip install -r requirements.txt` takes a few minutes.

## How to run a notebook-only module

Open the `*_explained.ipynb` (or `intro_ai_security.ipynb` / `guardrail_demo.ipynb`) inside that module's folder and run the cells top to bottom.

## How to run an app module (02, 04, 05, 06, 07, 08, 10)

```bash
# Terminal 1 - backend (if the module has one)
cd 0X-module-name
..\venv\Scripts\activate
uvicorn main:app --reload --port <see that module's README>    # 02 and 05 use backend.main:app

# Terminal 2 - frontend
cd 0X-module-name
..\venv\Scripts\activate
streamlit run app.py
```

Exact port numbers and file names are in each module's own `README.md`.
