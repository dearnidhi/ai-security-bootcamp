# Agent Security & AgentOps — Explained Simply

This notebook explains two related ideas in plain English: **Agent Security** (the risks) and **AgentOps** (how to watch for them). No code needed to follow along — the working demo is in `app.py` / `backend/main.py` in this same folder.

## What is an AI "agent"?

A normal chatbot just replies with text. An **agent** is an AI that can also *take actions* — it's given tools, like:

- Looking up a customer's order or invoice
- Sending an email
- Browsing a webpage
- Booking something, placing an order, updating a database

The AI decides, on its own, *which* tool to call, *when*, and with *what arguments*. That decision-making is exactly what makes agents powerful — and exactly what makes them risky.

## What is Agent Security?

Agent Security is about making sure an agent only does what it's actually supposed to do, even when someone tries to trick it.

**The core risk is called "Excessive Agency"** — when an agent is given more power than it needs, or isn't double-checked before it acts. If the agent gets tricked (by a malicious message, a poisoned webpage, anything), the damage it can cause is exactly as big as the power it was given.

**Example from this folder's demo:** a support agent has a `send_email` tool, meant to email customers their own invoices. A fake support ticket asks it to email the invoice to a *different* address instead. With no check in place, the agent just does it — leaking the customer's card details to a stranger.

## What is AgentOps?

AgentOps is short for "Agent Operations" — it means **watching what an agent actually does**, step by step: which tool it called, what arguments it used, and what came back.

Think of it like DevOps for a normal app — you keep logs and metrics so you know what's happening. With agents, this matters even more, because the agent itself decides what to do next. "It gave the right final answer" isn't enough information — you need to see *every step it took to get there*, or you can't tell when something went wrong (or why).

## How this folder's demo shows both ideas

Open `backend/main.py` in this folder. Every time the agent calls a tool, this line runs:

```python
trace.append({"tool": tool_call.function.name, "arguments": args, "result": result})
```

That `trace` list is a tiny, hand-rolled version of AgentOps — a record of every action the agent took. The Streamlit UI (`app.py`) displays this trace so you can see *exactly* when the agent decided to email the wrong address, and why.

Without that trace, all you'd see is "email sent" — with no way to know it went to an attacker.

## Real tools that do this in production

Companies don't usually hand-roll a trace list like this demo does — they use dedicated platforms:

- **AgentOps** — a platform (and the origin of the term) for tracing and monitoring AI agents
- **LangSmith** — built by the LangChain team, traces agent/LLM calls in detail
- **Arize Phoenix** — open-source observability for AI agents and LLM apps

These give you a dashboard instead of a printed list — but the underlying idea is exactly the same as the `trace` list in this demo.
